```python
import os
from dotenv import load_dotenv

load_dotenv()
```

```python
if not os.getenv("GROQ_API_KEY"):
    raise ValueError("Set GROQ_API_KEY in your environment/.env before running this notebook.")
```

```python
if not os.getenv("GOOGLE_API_KEY"):
    raise ValueError("Set GOOGLE_API_KEY in your environment/.env before running this notebook.")
```

```python
import os
os.environ["MEM0_TELEMETRY"] = "False"
```

```python
mem0_config = {
    "llm": {
        "provider": "gemini",
        "config": {
            "model": "gemini-3.1-flash-lite",
            "api_key": os.environ["GOOGLE_API_KEY"],
            "temperature": 0,
            "max_tokens":2000,
        },
    },
    "embedder": {
        "provider": "gemini",
        "config": {
            "model": "models/gemini-embedding-001",
            "embedding_dims": 1536,
            "api_key": os.environ["GOOGLE_API_KEY"],
        },
    },
    "vector_store": {
        "provider": "qdrant",
        "config": {
            "path": ":memory:",
            "embedding_model_dims": 1536,
        },
    }
}
```

Initialize Memory

```python
from mem0 import Memory

m = Memory.from_config(mem0_config)
```

Add a memory

```python
messages = [
    {"role": "user", "content": "Hi, I'm Alex. I love basketball and gaming."},
    {"role": "assistant", "content": "Hey Alex! I'll remember your interests."}
]
m.add(messages, user_id="alex")
```

Search memories

```python
results = m.search("What do you know about me?", filters={"user_id": "alex"})
print(results)
```

add new memory

```python
messages = [
    {"role": "user", "content": "Hi, I'm Bhargab. I love basketball and art."},
    {"role": "assistant", "content": "Hey Alex! I'll remember your interests."}
]
m.add(messages, user_id="Bhargab")
```
